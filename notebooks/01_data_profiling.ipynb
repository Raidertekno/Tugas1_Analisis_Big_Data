# Milestone 1 — Data Profiling

## Dataset Berita Politik Indonesia

Dataset yang digunakan merupakan dataset berita politik Indonesia dalam format Parquet.
Profiling dilakukan menggunakan Polars untuk mengetahui struktur, ukuran, kualitas,
dan karakteristik awal dataset sebelum dilakukan proses cleaning.

```python
import polars as pl
import glob
import os

print("Library berhasil di-import!")
print("Versi Polars:", pl.__version__)
```

```python
files = sorted(glob.glob("../data/*.parquet"))

print("Jumlah file:", len(files))

for file in files:
    print("-", os.path.basename(file))
```

```python
lf = pl.scan_parquet("../data/*.parquet")

print("Dataset berhasil di-scan menggunakan Polars LazyFrame.")
```

```python
schema = lf.collect_schema()

print("=== SCHEMA DATASET ===")

for column, dtype in schema.items():
    print(f"{column}: {dtype}")
```

```python
row_count = lf.select(
    pl.len().alias("jumlah_baris")
).collect()

print(row_count)
```

```python
column_count = len(schema)

print("Jumlah kolom:", column_count)
```

## 1. Ukuran Dataset

Bagian ini digunakan untuk memverifikasi jumlah baris dan kolom dataset
serta memastikan dataset memenuhi ketentuan minimal ukuran yang ditetapkan.

```python
print("Jumlah file   :", len(files))
print("Jumlah baris  :", row_count.item())
print("Jumlah kolom  :", column_count)
```

```python
preview = lf.head(5).collect()

preview
```

```python
last_data = lf.tail(5).collect()

last_data
```

```python
missing = (
    lf
    .null_count()
    .collect()
    .transpose(
        include_header=True,
        header_name="kolom",
        column_names=["jumlah_null"]
    )
)

missing
```

```python
missing_percentage = (
    missing
    .with_columns(
        (
            pl.col("jumlah_null") / row_count.item() * 100
        )
        .round(2)
        .alias("persentase_null")
    )
)

missing_percentage
```

```python
duplicate_count = (
    lf
    .select(
        pl.struct(lf.collect_schema().names())
        .is_duplicated()
        .sum()
        .alias("jumlah_duplicate")
    )
    .collect()
)

duplicate_count
```

```python
jumlah_sumber = (
    lf
    .select(
        pl.col("news_source").n_unique().alias("jumlah_sumber_unik")
    )
    .collect()
)

jumlah_sumber
```

```python
jumlah_wilayah = (
    lf
    .select(
        pl.col("news_kabkot").n_unique().alias("jumlah_wilayah_unik")
    )
    .collect()
)

jumlah_wilayah
```

```python
wilayah = (
    lf
    .select("news_kabkot")
    .unique()
    .head(20)
    .collect()
)

wilayah
```

```python
tanggal = (
    lf
    .select("news_date")
    .head(10)
    .collect()
)

tanggal
```

```python
tanggal_range = (
    lf
    .select(
        pl.col("news_date").min().alias("tanggal_min"),
        pl.col("news_date").max().alias("tanggal_max")
    )
    .collect()
)

tanggal_range
```

```python
top_sumber = (
    lf
    .group_by("news_hostname")
    .agg(
        pl.len().alias("jumlah_berita")
    )
    .sort("jumlah_berita", descending=True)
    .head(10)
    .collect()
)

top_sumber
```

```python
top_wilayah = (
    lf
    .group_by("news_kabkot")
    .agg(
        pl.len().alias("jumlah_berita")
    )
    .sort("jumlah_berita", descending=True)
    .head(10)
    .collect()
)

top_wilayah
```

## Kesimpulan Profiling

Berdasarkan proses profiling awal, dataset memiliki jumlah baris yang
memenuhi ketentuan minimal dataset pada tugas. Dataset terdiri dari
15 atribut yang memuat informasi mengenai berita, sumber berita,
tanggal publikasi, isi berita, serta wilayah kabupaten/kota.

Kolom `news_date` dapat digunakan sebagai atribut temporal, sedangkan
`news_kabkot` dapat digunakan sebagai atribut spasial.

Hasil pemeriksaan missing value, duplikasi, tipe data, distribusi sumber
berita, dan distribusi wilayah menjadi dasar untuk menentukan proses
cleaning pada Milestone 2.
